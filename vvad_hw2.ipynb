```python
import random
import time
from typing import List

def merge(arr: List[int], left: int, mid: int, right: int) -> None:
    n1 = mid - left + 1
    n2 = right - mid

    left_arr = [0] * n1
    right_arr = [0] * n2

    for i in range(n1):
        left_arr[i] = arr[left + i]
    for i in range(n2):
        right_arr[i] = arr[mid + 1 + i]

    i = j = 0
    k = left

    while i < n1 and j < n2:
        if left_arr[i] <= right_arr[j]:
            arr[k] = left_arr[i]
            i += 1
        else:
            arr[k] = right_arr[j]
            j += 1
        k += 1

    while i < n1:
        arr[k] = left_arr[i]
        i += 1
        k += 1

    while j < n2:
        arr[k] = right_arr[j]
        j += 1
        k += 1

def merge_sort(arr: List[int], left: int, right: int) -> None:
    if left < right:
        mid = left + (right - left) // 2

        merge_sort(arr, left, mid)
        merge_sort(arr, mid + 1, right)

        merge(arr, left, mid, right)

def generate_random_array(size: int, low: int, high: int) -> List[int]:
    return [random.randint(low, high) for _ in range(size)]

def print_array(arr: List[int]) -> None:
    print(" ".join(map(str, arr)))

def main() -> None:
    random.seed(time.time())

    size = 20
    low = 0
    high = 100

    arr = generate_random_array(size, low, high)

    print("Исходный массив:")
    print_array(arr)

    merge_sort(arr, 0, len(arr) - 1)

    print("Отсортированный массив:")
    print_array(arr)

if __name__ == "__main__":
    main()
```

Ниже приведен код для построения 3D-графика функции с помощью Plotly.

```python
import plotly.graph_objects as go


def plot_3d_surface():
    """
    Функция для создания и отображения 3D-графика поверхности, 
    которая представляет собой косинус от расстояния до начала координат.

    Параметры:
    -----------
    Нет входных параметров.

    Возвращает:
    -----------
    Нет возвращаемого значения. Функция отображает интерактивный 3D-график.
    """

    # Создание массива значений от -10 до 10 с 200 точками по оси X и Y
    x = np.linspace(-10, 10, 200)
    y = np.linspace(-10, 10, 200)

    # Создание сетки координат (X, Y) с помощью meshgrid
    x, y = np.meshgrid(x, y)

    # Вычисление значений Z как косинус от расстояния до начала координат (sqrt(x^2 + y^2))
    z = np.cos(np.sqrt(x**2 + y**2))

    # Создание 3D-графика поверхности с использованием Plotly
    fig = go.Figure(data=[go.Surface(z=z, x=x, y=y)])

    # Настройка макета графика
    fig.update_layout(
        title='Cosine Function 3D Surface Plot',  # Заголовок графика
        scene=dict(
            xaxis_title='X Axis',  # Подпись оси X
            yaxis_title='Y Axis',  # Подпись оси Y
            zaxis_title='Z Axis',  # Подпись оси Z
            aspectmode='cube',  # Режим отображения осей (кубический)
        ),
        margin=dict(l=0, r=0, b=0, t=0)  # Установка отступов вокруг графика
    )

    # Отображение графика
    # fig.show()

    # сохраним график в хтмл файл
    fig.write_html("hw_2_first_plot.html")

# Вызов функции для построения и отображения 3D-графика в файл хтмл
plot_3d_surface()
```

```python
#hw_2_third_plot

def plot_3d_surface(**kwargs):
    """
    Функция для создания и отображения 3D-графика поверхности с гибкими настройками.

    Параметры:
    -----------
    **kwargs : dict
        Параметры для настройки графика. Могут включать:
        - layout_params: параметры для update_layout (например, title, margin, scene и т.д.).
        - surface_params: параметры для go.Surface (например, colorscale, opacity, lighting и т.д.).

    Возвращает:
    -----------
    Нет возвращаемого значения. Функция отображает интерактивный 3D-график.
    """
    # Создание данных
    x = np.linspace(-10, 10, 200)
    y = np.linspace(-10, 10, 200)
    x, y = np.meshgrid(x, y)
    z = np.cos(np.sqrt(x**2 + y**2))

    # Параметры для поверхности (по умолчанию)
    surface_params = {
        'z': z,
        'x': x,
        'y': y,
        'colorscale': [[0.0, 'rgb(255, 182, 193)'], [1.0, 'rgb(173, 216, 230)']],  # Светло-розовый -> светло-голубой
    }

    # Параметры для макета (по умолчанию)
    layout_params = {
        'title': 'Cosine Function 3D Surface Plot',
        'scene': {
            'xaxis_title': 'X Axis',
            'yaxis_title': 'Y Axis',
            'zaxis_title': 'Z Axis',
            'aspectmode': 'cube',
        },
        'margin': dict(l=0, r=0, b=0, t=0),
    }

    # Обновление параметров поверхности и макета из kwargs
    if 'surface_params' in kwargs:
        surface_params.update(kwargs['surface_params'])
    if 'layout_params' in kwargs:
        layout_params.update(kwargs['layout_params'])

    # Создание 3D-графика
    fig = go.Figure(data=[go.Surface(**surface_params)])

    # Применение параметров макета
    fig.update_layout(**layout_params)

    # Отображение графика
    # fig.show()

    # сохраним график в хтмл файл
    fig.write_html("hw_2_third_plot.html")

# Пример вызова функции с пользовательскими параметрами
plot_3d_surface(
    surface_params={
        'colorscale': 'Viridis',  # Используем встроенную палитру
        'opacity': 0.8,  # Прозрачность поверхности
    },
    layout_params={
        'title': 'Custom 3D Surface Plot',
        'scene': {
            'xaxis_title': 'Custom X Axis',
            'yaxis_title': 'Custom Y Axis',
            'zaxis_title': 'Custom Z Axis',
        },
    }
)
```

**Инструкция:**
Функция plot_3d_surface предназначена для создания и отображения 3D-графика поверхности с гибкими настройками. Она позволяет пользователю передавать параметры для настройки как самой поверхности, так и макета графика через аргументы **kwargs. Ниже приведено подробное описание работы функции и инструкция по её использованию.
1. Назначение функции

    Функция создает 3D-график поверхности, где:

        Оси X и Y задают координаты на плоскости.

        Ось Z вычисляется как косинус от расстояния до центра (функция z = np.cos(np.sqrt(x**2 + y**2))).

        График можно гибко настраивать с помощью параметров, передаваемых через **kwargs.

3. Параметры функции

    Функция принимает следующие параметры через **kwargs:

        surface_params (опционально):

            Словарь с параметрами для настройки поверхности (go.Surface).

        layout_params (опционально):

            Словарь с параметрами для настройки макета графика (update_layout).

    Создание данных:

        Генерируются массивы x и y с помощью np.linspace, которые задают диапазон значений по осям X и Y.

        Создается сетка координат с помощью np.meshgrid.

        Вычисляются значения z как косинус от расстояния до центра.

    Параметры по умолчанию:

        Для поверхности (surface_params) и макета (layout_params) задаются значения по умолчанию.

    Обновление параметров:

        Если пользователь передает свои параметры через kwargs, они обновляют значения по умолчанию с помощью метода update.

    Создание графика:

        Создается объект go.Figure с использованием параметров поверхности.

        Применяются параметры макета с помощью update_layout.

    Отображение графика:

        График отображается с помощью fig.show().

7. Список поддерживаемых параметров

    Для surface_params:

        colorscale: Цветовая палитра (например, 'Viridis', 'Plasma', или пользовательская палитра).

        opacity: Прозрачность поверхности (от 0.0 до 1.0).

        lighting: Настройки освещения (например, dict(ambient=0.7)).

        showscale: Показывать ли шкалу цветов (по умолчанию True).

    Для layout_params:

        title: Заголовок графика.

    # Формальная постановка задачи

    Дано: функция $$f(x)$$ в виде дерева с операторами $\{+, -, \times, \div, \sin, \cos, \log, \exp\}$ и листьями $\{x, \mathbb{Z}, \pi, e\}$  
    Ограничения:
    - Глубина дерева $\leq 15$ операторов  
    - Коэффициенты $c_i \in [-10, 10] \cap \mathbb{Z} \setminus \{0\}$  
    Задача:  
        Найти $F(x)$ такую, что:  
        $$\frac{dF}{dx} = f(x)$$  
    Решение ОДУ
        ОДУ 1-го порядка
        $$\frac{dy}{dx} + P(x)y = Q(x), \quad y(x_0) = y_0$$
        
        ОДУ 2-го порядка  
        $$y'' + p(x)y' + q(x)y = r(x), \quad y(x_0) = y_0, \ y'(x_0) = y'_0$$
    Представление данных
    - Вход: Дерево $\mathcal{T}$ → последовательность токенов (префиксная нотация)  
    - Выход:  
      - Для интегралов: $\mathcal{T}_{F}$  
      - Для ОДУ: $\mathcal{T}_{y}$  
    Проверка:  
    - Интегралы: $\frac{d}{dx}\mathcal{T}_{F} \equiv \mathcal{T}_f$  
    - ОДУ: Подстановка $\mathcal{T}_{y}$ в исходное уравнение  
    Ограничения
        1. Макс. сложность: 15 операторов  
        2. Запрещены:  
           - Неопределенные интегралы без границ  
           - Особые точки ($\log(0)$, деление на 0)  
        3. Методы генерации: FWD, BWD, IBP  
    
    Метрики
    - Accuracy: Доля верных решений  
    - Валидация: SymPy + сравнение с Mathematica/MATLAB  

    Структура входных данных.
        1. Тип данных. Данные представлены на входе в виде функций, дифференциальных и интегральных уравнений, которые формализованы как 
          последовательности символов.
        2. Формат. Авторы используют деревья для представления математических выражений, но для обучения модели эти деревья преобразуются в 
          последовательности с использованием префиксной нотации.
          
    Ограничения на данные.
        1. Ограничения на сложность: Не более 15 операторов в выражении. 
        2. Допустимые диапазоны чисел: Целые числа от −10 до 10 (кроме 00), стандартные константы.
        3. Пропущенные значения: Не допускаются.
        4. Некорректные выражения: Исключаются.
        5. Баланс данных: Обеспечивается за счёт комбинации методов генерации и контроля длины выражений.
        6. Уникальность и упрощение: Удаляются дубликаты и упрощаются выражения.

            a) Общий тестовый набор
