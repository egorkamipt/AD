Для $U[0, 1]$ сгенерируйте $200$ независимых случайных величин, постройте график плотности на отрезке $[-0.25, 1.25]$, а также гистограмму по сгенерированным случайным величинам. Для построения графика необходимо брать достаточно большое количество точек сетки.

```python
sample = sps.uniform.rvs(size=200)  # выборка размера 200
grid = np.linspace(-0.25, 1.25, 200)  # сетка для построения графика

plt.figure(figsize=(12, 5))
plt.hist(sample, bins=30, density=True,
         alpha=0.6, label='Плотность значений выборки')
plt.plot(grid, sps.uniform.pdf(grid), color='red',
         lw=5, label='Плотность случайной величины')
plt.title(r'Случайная величина $\xi \sim U$[0, 1]', fontsize=20)
plt.legend(fontsize=14, loc=1)
plt.show()
```

Исследуйте, как меняются значения случайных величин в зависимости от `precision` в написанной ранее вами функции.

```python
# функция coin
def coin(shape):
    return sps.bernoulli.rvs(p=0.5, size=shape)

# функция uniform
def uniform(size=1, precision=30):
    shape = (size, precision) if isinstance(size, int) else tuple(size) + (precision,)
    return coin(shape) @ np.geomspace(0.5, 0.5 ** precision, precision)
```

```python
size = 1000
precisions = [1, 2, 3, 4, 5, 6]

plt.figure(figsize=(15, 10))
for i, precision in enumerate(precisions):
    plt.subplot(3, 2, i + 1)
    plt.xlim(0, 1)
    plt.scatter(uniform(size, precision), np.zeros(size))
    plt.title(f'Precision: {precision}')
    plt.yticks([])

plt.tight_layout()
plt.show()
```

Из графиков видно, что с увеличением precision увеличивается колчество значений случайных величин.

$$p(x) = \frac{1}{\pi(1+x^2)}$$
для $x\in\mathbb{R}$, которое известно своими <a target="_black" href="https://en.wikipedia.org/wiki/Heavy-tailed_distribution">тяжелыми хвостами</a>. **1.** Визуализируйте на одном графики плотности нормального распределения и распределения Коши. Что можно сказать?

### Задача 3.

Перед выполнением этой задачи настоятельно рекомендуется посмотреть <a href="https://thetahat.ru/courses/ad/main/9/lec9_LLN" target="_top">ноутбук</a> с лекции про закон больших чисел.

На практике не редко возникает вопрос &mdash; **сколько времени потребуется на сбор данных, чтобы оценить среднее с достаточной точностью**. Для ответа на этот вопрос можно использовать различные теоретические оценки, в частности, неравенство Чебышева.

Другой способ &mdash; смоделировать большое количество выборок.
Возьмем в качестве меры отклонения от среднего *среднее значение модуля отклонения от среднего*, то есть
$$F(n) = \frac{1}{K}\sum\limits_{j=1}^{K} \left| S_{nj} - a \right|,$$
где $a$ &mdash; математическое ожидание элемента выборки, $S_{nj}$ &mdash; *выборочное* среднее $n$ элементов в $j$-й выборке, а $K$ &mdash; общее количество выборок.

**1.** Рассмотрим распределения $Bern(a)$ для $a=0.5$. Пусть для нас допустимо отклонение на 0.02, иначе говоря при *истинном* среднем 0.5 мы считаем достаточно точными значения *выборочного* среднего от 0.48 до 0.52. Тогда в качестве *приближения* необходимого размера выборок можно взять наименьшее значение $n$, при котором $F(n) \leqslant 0.02$.

Реализуйте данный эксперимен для $K=500$  и найдите значение $n$. Визуализируйте зависимость $F(n)$ от $n$ в обычном и логарифмическом масштабах (по тем осям, по которым это имеет смысл).

Для реализации без циклов стоит генерировать достаточно большие выборки и использовать функцию `np.where`.

```python
size = 1000
samples_count = 500
a = 0.5
deviation = 0.02

samples = sps.bernoulli(p=a).rvs((samples_count, size))
cum_means = samples.cumsum(axis=1) / (np.arange(size) + 1)

F = np.mean(np.abs(cum_means - a), axis=0)
print("n =", len(np.where(F > deviation)[0]))

plt.figure(figsize=(15, 5))
plt.plot(F, lw=3)
plt.hlines(deviation, 0, size, color="red")
plt.title('Зависимость F(n) от n')
plt.xlabel('n')
plt.ylabel('F(n)')
plt.show()
```

```python
plt.figure(figsize=(15, 5))
plt.plot(np.log(F), lw=3)
plt.hlines(np.log(deviation), 0, size, color="red")
plt.title('Зависимость F(n) от n в логарифмическом масштабе')
plt.xlabel('n')
plt.ylabel('ln(F(n))')
plt.show()
```

**2.** Повторите данный эксперимент для значений $a$ от 0.05 до 0.95 с шагом 0.05. Визуализируйте зависимость необходимого размера выборки $n$ от значения $a$.

При реализации можно сделать цикл по значениям $a$.

```python
size = 1000
samples_count = 500
deviation = 0.02

n_a = []
for a in np.arange(0.05, 1.0, 0.05):
    samples = sps.bernoulli(p=a).rvs((samples_count, size))
    cum_means = samples.cumsum(axis=1) / (np.arange(size) + 1)

    F = np.mean(np.abs(cum_means - a), axis=0)

    n_a.append(len(np.where(F >= deviation)[0]))

plt.figure()
plt.plot(np.arange(0.05, 1.0, 0.05), n_a)
plt.title('Зависимость n от a')
plt.xlabel('a')
plt.ylabel('n')
plt.show()
```

**Выводы:**

F(n) попадает в ворота с n порядка сотен, это неплохой результат.
Последний график иммметричен относительно 0.5 и имеет максимум в этой точке. Это предсказывается теорией, действительно: дисперсия Бернулли p(1-p) максимальна при p = 0.5

Рассмотрим следующее вероятностное пространство:
* $\Omega = \{1, 2, 3, 4, 5, 6, 7, 8\}$ – номера точек на единичной окружности.

download (1) (1).png
* $\sigma$-алгебра $\mathscr{F}$, которая включает все подмножества множества $\Omega$, $\mathscr{F} = 2^{\{1, 2, 3, 4, 5, 6, 7, 8\}}$
* $\mathsf{P}$ - равномерное распределение:
$$
\mathsf{P}(\omega = \{k\}) = \frac{1}{8}, \quad k = 1, 2, \ldots, 8.
$$

Определим случайную величину следующим образом:
$$
X(\omega) = \sin\left(\frac{2\pi (\omega - 1)}{8}\right) = \sin\left(\frac{\pi (\omega - 1)}{4}\right)$$
Интерпретация: значение координаты $y$ для точки с номером $w$ на единичной окружности.


Пусть случайная величина $Y$ определяется следующим образом:
$$ Y(\omega) = \begin{cases}
1, & \text{если } \sin\left(\frac{\pi (\omega - 1)}{4}\right) \geq 0, \\
0, & \text{иначе.}
\end{cases} $$


Какому наблюдению соответствует случайная величина $Y$?

Её можно поставить в соответствие бинаризации некоторого категориального признака.



Найдите:
* Условное распределение $X|Y$;
* Условное математическое ожидание $\mathsf{E}(X|Y=1)$;
* Условное математическое ожидание $\mathsf{E}(X|Y=0)$.

Визуализируйте условное распределение $X|Y$.

$\textbf{Условное распределение \( X|Y \)}$

При $ Y = 1 $:
$$
X|Y=1 \sim 
\begin{cases}
0, & \text{с вероятностью } \dfrac{2}{5}, \\
\dfrac{\sqrt{2}}{2}, & \text{с вероятностью } \dfrac{2}{5}, \\
1, & \text{с вероятностью } \dfrac{1}{5}.
\end{cases}
$$

При $Y = 0$:
$$
X|Y=0 \sim 
\begin{cases}
-\dfrac{\sqrt{2}}{2}, & \text{с вероятностью } \dfrac{2}{3}, \\
-1, & \text{с вероятностью } \dfrac{1}{3}.
\end{cases}
$$

$\textbf{Условные математические ожидания}$

$$
\mathsf{E}(X|Y=1) = \dfrac{1 + \sqrt{2}}{5} \approx 0.483,
$$
$$
\mathsf{E}(X|Y=0) = -\dfrac{1 + \sqrt{2}}{3} \approx -0.805.
$$

```python

# Y=1
x_y1 = [0, np.sqrt(2)/2, 1]
prob_y1 = [2/5, 2/5, 1/5]

# Y=0
x_y0 = [-np.sqrt(2)/2, -1]
prob_y0 = [2/3, 1/3]

# Создание фигуры
plt.figure(figsize=(10, 5))

# Точечный график для Y=1 (верхние точки)
plt.scatter(x_y1, prob_y1, s=[p*1000 for p in prob_y1], 
           color='violet', alpha=0.7, label='Y=1', zorder=3)

# Точечный график для Y=0 (нижние точки)
plt.scatter(x_y0, prob_y0, s=[p*1000 for p in prob_y0], 
           color='orange', alpha=0.7, label='Y=0', zorder=3)

# Аннотации вероятностей
for x, p, color in zip(x_y1 + x_y0, prob_y1 + prob_y0, 
                      ['violet']*3 + ['orange']*2):
    plt.text(x, p + 0.02, f'{p:.2f}', ha='center', color=color, fontsize=10)

# Настройка осей и легенды
plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
plt.xlabel('Значение X', fontsize=12)
plt.ylabel('Вероятность', fontsize=12)
plt.title('Условное распределение X|Y', fontsize=14)
plt.xticks([-1, -np.sqrt(2)/2, 0, np.sqrt(2)/2, 1], 
          ['-1', r'$-\frac{\sqrt{2}}{2}$', '0', r'$\frac{\sqrt{2}}{2}$', '1'])
plt.yticks(np.arange(0, 0.8, 0.1))
plt.ylim(-0.05, 0.75)
plt.grid(alpha=0.3, linestyle=':')
plt.legend()
plt.show()
```
